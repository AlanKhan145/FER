### **Tiền xử lý dữ liệu**

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

def preprocess_image(image_path, save_path, target_size=(48, 48), margin=0.05):
    """
    1. Chuẩn hóa ảnh
    Hàm này thực hiện chuẩn hóa ảnh đầu vào từ khoảng [0, 255] sang [0, 1], sau đó chuyển sang khoảng [-1, 1].
    2. Tăng cường độ tương phản của ảnh (CLAHE)
    3. Phát hiện khuôn mặt
    Phát hiện khuôn mặt bằng thuật toán Haar Cascade và điều chỉnh vùng chứa khuôn mặt.
    4. Cắt ảnh khuôn mặt và chuẩn hóa
    Cắt ảnh khuôn mặt và thay đổi kích thước ảnh về kích thước chuẩn, sau đó chuẩn hóa ảnh vào khoảng [-1, 1].
    5. Lưu ảnh đã xử lý
    Lưu ảnh đã chuẩn hóa vào đường dẫn chỉ định và trả về ảnh đã chuẩn hóa.
    """

    # Tải mô hình Haar Cascade được cung cấp bởi OpenCV
    haar_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_default.xml")

    # Đọc ảnh từ đường dẫn
    image = cv2.imread(image_path)
    if image is None:
        print(f"Không thể tải ảnh '{image_path}'.")
        return

    # Giảm kích thước ảnh để giảm độ phức tạp
    image = cv2.resize(image, (image.shape[1] // 2, image.shape[0] // 2))

    # Chuyển đổi ảnh sang màu xám để tăng cường độ tương phản
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # 2. Tăng cường độ tương phản của ảnh (CLAHE)
    clahe = cv2.createCLAHE(clipLimit=3.0, tileGridSize=(8, 8))
    enhanced_image = clahe.apply(gray)

    # Phát hiện khuôn mặt
    faces = haar_cascade.detectMultiScale(enhanced_image, scaleFactor=1.1, minNeighbors=5, minSize=(30, 30))

    if len(faces) > 0:
        # Lấy khuôn mặt đầu tiên
        x, y, w, h = faces[0]

        # Tính margin để điều chỉnh tọa độ
        x_margin = int(w * margin)
        y_margin = int(h * margin)

        # Điều chỉnh tọa độ để lấy từ mắt đến miệng (bỏ cằm)
        x = max(0, x - x_margin)
        x2 = min(image.shape[1], x + w + 2 * x_margin)

        # Chỉ lấy phần từ mắt đến miệng, không bao gồm cằm
        y_eye_to_mouth_start = int(y + 0.3 * h)  # 30% từ trên xuống (bắt đầu từ mắt)
        y_eye_to_mouth_end = int(y + 0.8 * h)   # 80% từ trên xuống (kết thúc trước cằm)

        y = max(0, y_eye_to_mouth_start - y_margin)
        y2 = min(image.shape[0], y_eye_to_mouth_end + y_margin)

        # Cắt phần ảnh từ mắt đến miệng (không có cằm)
        face_image = enhanced_image[y:y2, x:x2]
    else:
        print(f"Không tìm thấy khuôn mặt trong ảnh '{image_path}'. Sử dụng toàn bộ ảnh.")
        face_image = enhanced_image  # Nếu không tìm thấy khuôn mặt, sử dụng ảnh gốc

    # Thay đổi kích thước ảnh về (48, 48)
    resized_image = cv2.resize(face_image, target_size)

    # Chuyển đổi ảnh từ grayscale sang RGB (3 kênh)
    rgb_image = cv2.cvtColor(resized_image, cv2.COLOR_GRAY2RGB)

    # Chuẩn hóa ảnh vào khoảng [-1, 1]
    normalized_image = rgb_image / 127.5 - 1.0

    # Lưu ảnh đã chuẩn hóa (lưu dưới dạng giá trị từ 0-255 để trực quan)
    save_image = ((normalized_image + 1) * 127.5).astype(np.uint8)
    cv2.imwrite(save_path, save_image)

    print(f"Đã xử lý và lưu ảnh vào '{save_path}'.")

    return normalized_image  # Trả về ảnh đã chuẩn hóa để sử dụng tiếp (nếu cần)

# Sử dụng hàm với một ảnh cụ thể
raw_path = "/content/raw.jpg"
save_path = "/content/preprocessed_eye_to_mouth_no_chin.jpg"
normalized_image = preprocess_image(raw_path, save_path, margin=0.05)

# Kiểm tra và hiển thị ảnh đã xử lý
if normalized_image is not None:
    plt.imshow(normalized_image, cmap='gray')
    plt.axis('off')
    plt.title("Ảnh đã chuẩn hóa [-1, 1]")
    plt.show()
```

```python
import os
import cv2
import random


# Hàm để xử lý tất cả các ảnh trong một thư mục và lưu vào một thư mục đích
def load_and_preprocess_images(directory_path, save_directory_path, target_size=(48, 48)):
    """
    Chỉnh lại kích thước của ảnh trong thư mục đầu vào và lưu vào thư mục đích.
    :param directory_path: Đường dẫn đến thư mục chứa ảnh gốc.
    :param save_directory_path: Đường dẫn đến thư mục để lưu các ảnh đã chỉnh kích thước.
    :param target_size: Kích thước mục tiêu của ảnh dưới dạng tuple (chiều cao, chiều rộng).
    """
    # Kiểm tra nếu thư mục lưu không tồn tại, nếu không, tạo mới
    if not os.path.exists(save_directory_path):
        os.makedirs(save_directory_path)

    # Lấy danh sách tệp ảnh
    image_files = [f for f in os.listdir(directory_path) if f.endswith(('.jpg', '.jpeg', '.png', '.bmp', '.gif'))]

    # Lặp qua các tệp ảnh đã chọn
    for filename in image_files:
        image_path = os.path.join(directory_path, filename)
        save_file_path = os.path.join(save_directory_path, filename)
        preprocess_image(image_path, save_file_path, target_size)
```

```python
import os
import shutil

def get_dataset_info(directory):
    dataset_info = {}
    total_files = 0

    if not os.path.exists(directory):
        return f"Thư mục không tồn tại: {directory}"

    # Duyệt qua từng thư mục con trong thư mục gốc
    for subfolder in os.listdir(directory):
        subfolder_path = os.path.join(directory, subfolder)
        if os.path.isdir(subfolder_path):
            # Đếm số lượng tệp trong từng thư mục con
            num_files = len([f for f in os.listdir(subfolder_path) if os.path.isfile(os.path.join(subfolder_path, f))])
            dataset_info[subfolder] = num_files
            total_files += num_files

    # Tổng quan
    dataset_info['total_classes'] = len(dataset_info)
    dataset_info['total_files'] = total_files
    return dataset_info


```

```python
# Đường dẫn đến các thư mục
train_dir = '/content/data/train'
validation_dir = '/content/data/validation'
test_dir = '/content/data/test'

# Lấy thông tin bộ dữ liệu
train_info = get_dataset_info(train_dir)
validation_info = get_dataset_info(validation_dir)
test_info = get_dataset_info(test_dir)

# In thông tin
print("Thông tin bộ dữ liệu Train:")
print(train_info)
print("\nThông tin bộ dữ liệu Validation:")
print(validation_info)
print("\nThông tin bộ dữ liệu Test:")
print(test_info)

```

```python
import tensorflow as tf
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Dense, Dropout, GlobalAveragePooling2D, BatchNormalization
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import ReduceLROnPlateau
from tensorflow.keras import regularizers
from keras.losses import CategoricalFocalCrossentropy

def create_emotion_classifier_mobilenetv2(input_shape=(48, 48, 3), num_classes=7):
    """
    Tạo mô hình phân loại cảm xúc sử dụng MobileNetV2 làm mô hình cơ sở với độ phức tạp cao hơn.
    """
    # Tải MobileNetV2 với trọng số đã được huấn luyện sẵn từ ImageNet, loại bỏ lớp cuối cùng
    base_model = MobileNetV2(input_shape=input_shape, include_top=False, weights='imagenet')

    # Mở khóa tất cả các lớp để tinh chỉnh
    for layer in base_model.layers:
        layer.trainable = True

    # Thêm các lớp tùy chỉnh trên đầu mô hình cơ sở
    x = base_model.output
    x = GlobalAveragePooling2D()(x)
    x = Dense(1024, activation='relu', kernel_regularizer=regularizers.l2(0.01))(x)  # Lớp Dense với L2 Regularization
    x = BatchNormalization()(x)
    x = Dropout(0.5)(x)
    x = Dense(512, activation='relu')(x)  # Lớp Dense thứ hai
    x = BatchNormalization()(x)
    x = Dropout(0.5)(x)
    x = Dense(256, activation='relu')(x)  # Lớp Dense thứ ba
    x = BatchNormalization()(x)
    x = Dropout(0.5)(x)

    # Lớp đầu ra
    predictions = Dense(num_classes, activation='softmax')(x)

    # Tạo mô hình đầy đủ
    model = Model(inputs=base_model.input, outputs=predictions)

    # Biên dịch mô hình với Adam optimizer và learning rate scheduling
    model.compile(optimizer=Adam(learning_rate=0.0001),
              loss=CategoricalFocalCrossentropy(),
              metrics=['accuracy'])
    return model

if __name__ == "__main__":
    # Tạo mô hình
    model = create_emotion_classifier_mobilenetv2()

    # In ra thông tin mô hình
    model.summary()
```

```python
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (Conv2D, MaxPooling2D, Dense, Dropout, Flatten,
                                     BatchNormalization, GlobalAveragePooling2D, Input)
from tensorflow.keras.optimizers import Adam
from tensorflow.keras import regularizers


def create_emotion_classifier_clcm(input_shape=(48, 48, 3), num_classes=7):
    """
    Tạo mô hình phân loại cảm xúc với CLCM (Convolutional Layer Classification Model).
    """
    inputs = Input(shape=input_shape)
    x = BatchNormalization()(inputs)

    # Tầng Conv2D và MaxPooling2D 1
    x = Conv2D(32, (3, 3), activation='relu', padding='same')(x)
    x = BatchNormalization()(x)
    x = MaxPooling2D((2, 2))(x)

    # Tầng Conv2D và MaxPooling2D 2
    x = Conv2D(64, (3, 3), activation='relu', padding='same')(x)
    x = BatchNormalization()(x)
    x = MaxPooling2D((2, 2))(x)

    # Tầng Conv2D và MaxPooling2D 3
    x = Conv2D(128, (3, 3), activation='relu', padding='same')(x)
    x = BatchNormalization()(x)
    x = MaxPooling2D((2, 2))(x)

    # Tầng Conv2D và MaxPooling2D 4
    x = Conv2D(256, (3, 3), activation='relu', padding='same')(x)
    x = BatchNormalization()(x)
    x = MaxPooling2D((2, 2))(x)

    # Tầng Conv2D bổ sung để tăng cường khả năng học
    x = Conv2D(512, (3, 3), activation='relu', padding='same', kernel_regularizer=regularizers.l2(0.01))(x)
    x = BatchNormalization()(x)
    x = MaxPooling2D((2, 2))(x)

    # Tầng GlobalAveragePooling2D và các lớp Dense
    x = GlobalAveragePooling2D()(x)
    x = Dropout(0.5)(x)  # Dropout bổ sung
    x = Dense(1024, activation='relu', kernel_regularizer=regularizers.l2(0.01))(x)
    x = BatchNormalization()(x)
    x = Dropout(0.5)(x)
    x = Dense(512, activation='relu', kernel_regularizer=regularizers.l2(0.01))(x)
    x = BatchNormalization()(x)
    x = Dropout(0.5)(x)
    x = Dense(256, activation='relu')(x)
    x = BatchNormalization()(x)
    x = Dropout(0.4)(x)

    # Lớp đầu ra với Softmax
    predictions = Dense(num_classes, activation='softmax')(x)

    model = Model(inputs=inputs, outputs=predictions)

    # Biên dịch mô hình với Adam optimizer và learning rate scheduling
    model.compile(optimizer=Adam(learning_rate=0.0001),
              loss=CategoricalFocalCrossentropy(),
              metrics=['accuracy'])
    return model


if __name__ == "__main__":
    # Tạo mô hình
    model = create_emotion_classifier_clcm()

    # In ra thông tin mô hình
    model.summary()

```

```python
import os
import cv2
import numpy as np
from tensorflow.keras.preprocessing.image import ImageDataGenerator

def augment_data(train_dir, batch_size=32, target_size=(48, 48), shuffle=True):
    """
    Áp dụng tăng cường dữ liệu cho bộ dữ liệu huấn luyện.
    :param train_dir: Thư mục chứa dữ liệu huấn luyện.
    :param batch_size: Số lượng ảnh xử lý cùng một lúc.
    :param target_size: Kích thước mục tiêu của ảnh.
    :param shuffle: Nếu True, dữ liệu sẽ được xáo trộn trước khi chia thành batch.
    :return: Đối tượng ImageDataGenerator của Keras.
    """
    # Khởi tạo ImageDataGenerator với các phép biến đổi tăng cường dữ liệu
    train_datagen = ImageDataGenerator(
        rescale=1./255,                # Chuẩn hóa giá trị pixel ảnh
        rotation_range=10,             # Xoay ảnh trong khoảng 10 độ
        width_shift_range=0.1,         # Dịch chuyển ảnh theo chiều rộng (10%)
        height_shift_range=0.1,        # Dịch chuyển ảnh theo chiều cao (10%)
        shear_range=0.1,               # Cắt xén ảnh để tạo độ nghiêng
        zoom_range=0.1,                # Phóng to ảnh
        horizontal_flip=True,          # Lật ngang ảnh
        fill_mode='nearest'            # Điền giá trị các pixel bị thiếu sau biến đổi
    )

    # Tạo bộ sinh dữ liệu từ thư mục huấn luyện và thêm tính năng xáo trộn
    train_generator = train_datagen.flow_from_directory(
        train_dir,
        target_size=target_size,        # Thay đổi kích thước ảnh
        batch_size=batch_size,          # Số lượng ảnh mỗi lần xử lý
        class_mode='categorical',       # Lớp phân loại ảnh
        shuffle=shuffle                 # Đảm bảo dữ liệu được xáo trộn nếu cần
    )

    return train_generator

def prepare_data(train_dir, val_dir, batch_size=128, target_size=(48, 48)):
    """
    Chuẩn bị dữ liệu cho huấn luyện và xác nhận (validation).
    :param train_dir: Thư mục chứa dữ liệu huấn luyện.
    :param val_dir: Thư mục chứa dữ liệu xác nhận (validation).
    :param batch_size: Số lượng ảnh xử lý cùng một lúc.
    :param target_size: Kích thước mục tiêu của ảnh.
    :return: Một tuple chứa đối tượng ImageDataGenerator của Keras cho huấn luyện và xác nhận.
    """
    # Áp dụng tăng cường dữ liệu cho bộ huấn luyện
    train_generator = augment_data(train_dir, batch_size, target_size, shuffle=True)

    # Tạo ImageDataGenerator cho dữ liệu xác nhận (validation), chỉ chuẩn hóa
    val_datagen = ImageDataGenerator(rescale=1./255)

    # Tạo bộ sinh dữ liệu cho dữ liệu xác nhận
    val_generator = val_datagen.flow_from_directory(
        val_dir,
        target_size=target_size,
        batch_size=batch_size,
        class_mode='categorical',
        shuffle=False  # Không xáo trộn dữ liệu xác nhận
    )

    return train_generator, val_generator



```

```python
import os
import pickle
import time
from tensorflow.keras.models import load_model
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping, ReduceLROnPlateau
from tensorflow.keras.optimizers import Adam
import tensorflow as tf

def load_existing_model(model_path):
    """Tải một model đã tồn tại từ một tệp .keras hoặc .h5."""
    return load_model(model_path)

def load_existing_history(history_save_path):
    """Tải lịch sử huấn luyện từ tệp pickle nếu tồn tại."""
    if os.path.exists(history_save_path):
        with open(history_save_path, 'rb') as f:
            history = pickle.load(f)
        print(f"Lịch sử huấn luyện đã được tải từ {history_save_path}")
        return history
    return None

def save_updated_history(history, history_save_path):
    """Lưu lịch sử huấn luyện đã được cập nhật vào tệp pickle."""
    with open(history_save_path, 'wb') as f:
        pickle.dump(history, f)
    print(f"Lịch sử huấn luyện đã được lưu tại {history_save_path}")


def train_mobilenetv2_model(train_dir, validation_dir, model_save_path, history_save_path, additional_epochs=10, batch_size=128):
    """Huấn luyện mô hình phân loại cảm xúc bằng MobileNetV2."""
    # Chuẩn bị dữ liệu
    train_generator, validation_generator = prepare_data(train_dir, validation_dir, batch_size)

    # Kiểm tra và tải model
    if not model_save_path:
        raise ValueError("model_save_path không được để trống.")

    model_dir = os.path.dirname(model_save_path) or '.'
    os.makedirs(model_dir, exist_ok=True)

    if os.path.exists(model_save_path):
        model = load_existing_model(model_save_path)
        print(f"Model đã tồn tại: {model_save_path}")
    else:
        print("Model chưa tồn tại. Tạo một model mới.")
        model = create_emotion_classifier_mobilenetv2(input_shape=(48, 48, 3), num_classes=7)


    # Kiểm tra lịch sử huấn luyện
    history = load_existing_history(history_save_path)
    initial_epoch = len(history.get('val_accuracy', [])) if history else 0

    if initial_epoch > 0:
        print(f"Tiếp tục huấn luyện từ epoch {initial_epoch + 1}")
    else:
        print("Bắt đầu huấn luyện từ epoch 0.")

    total_epochs = additional_epochs + initial_epoch

    # Tính toán các bước mỗi epoch
    steps_per_epoch = train_generator.samples // batch_size
    validation_steps = validation_generator.samples // batch_size

    # Khởi tạo callback
    checkpoint = ModelCheckpoint(model_save_path, monitor='val_accuracy',verbose=1)
    early_stopping = EarlyStopping(monitor='val_loss', patience=99, verbose=1)
    reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=3, min_lr=1e-6, verbose=1)

    # Khởi tạo thời gian huấn luyện
    start_time = time.time()

    # Huấn luyện mô hình
    new_history = model.fit(
        train_generator,
        steps_per_epoch=steps_per_epoch,
        epochs=total_epochs,
        validation_data=validation_generator,
        validation_steps=validation_steps,
        callbacks=[checkpoint, early_stopping, reduce_lr],
        initial_epoch=initial_epoch
    )

    # Kết thúc theo dõi thời gian
    training_time = time.time() - start_time

    # Tính toán tham số có thể đào tạo
    trainable_params = sum([tf.size(w).numpy() for w in model.trainable_weights])

    # Độ chính xác theo lớp
    class_accuracies = {
        class_name: accuracy
        for class_name, accuracy in zip(train_generator.class_indices.keys(), new_history.history.get('val_accuracy', []))
    }

    # In kết quả
    print(f"Độ Chính Xác Tổng Thể: {new_history.history.get('val_accuracy', [0])[-1] * 100:.2f}%")
    print(f"Độ Chính Xác Theo Lớp: {class_accuracies}")
    print(f"Tham Số Có Thể Đào Tạo: {trainable_params}")
    print(f"Thời Gian Huấn luyện: {training_time:.2f} giây")

    # Cập nhật lịch sử huấn luyện
    if history:
        for key in ['accuracy', 'val_accuracy', 'loss', 'val_loss']:
            history[key].extend(new_history.history.get(key, []))
    else:
        history = new_history.history

    # Lưu lịch sử huấn luyện đã cập nhật
    save_updated_history(history, history_save_path)

    return new_history

if __name__ == '__main__':
    train_dir = '/content/data/train'
    validation_dir = '/content/data/validation'

    try:
        # Huấn luyện thêm epoch sau khi đã học xong các epoch trước đó
        history = train_mobilenetv2_model(
            train_dir,
            validation_dir,
            model_save_path='/content/models/mobilenetv2.keras',
            history_save_path='/content/models/history.pkl',  # Đảm bảo đường dẫn chính xác
            additional_epochs=100,
            batch_size=128
        )
    except Exception as e:
        print(f"Có lỗi xảy ra: {e}")
```

```python
import os
import pickle
import matplotlib.pyplot as plt

def load_history(history_save_path):
    """
    Tải lịch sử huấn luyện từ tệp pickle nếu tồn tại.
    """
    if os.path.exists(history_save_path):
        with open(history_save_path, 'rb') as f:
            history = pickle.load(f)
        print(f"Lịch sử huấn luyện đã được tải từ {history_save_path}")
        return history
    else:
        print("Không tìm thấy tệp lịch sử huấn luyện.")
        return None

def plot_accuracy(history):
    """
    Vẽ biểu đồ độ chính xác của tập huấn luyện và tập kiểm thử cho các epoch lẻ.
    """
    if history is None:
        print("Không có lịch sử huấn luyện để vẽ biểu đồ.")
        return

    # Truy xuất độ chính xác của tập huấn luyện và tập kiểm thử
    train_acc = history.get('accuracy', [])
    val_acc = history.get('val_accuracy', [])

    # Kiểm tra xem có dữ liệu hay không
    if not train_acc or not val_acc:
        print("Không có độ chính xác của tập huấn luyện hoặc tập kiểm thử.")
        return

    # Lọc các epoch lẻ
    epochs = range(1, len(train_acc) + 1)
    odd_epochs = [epoch for epoch in epochs if epoch % 2 != 0]

    # Lọc độ chính xác tương ứng với các epoch lẻ
    odd_train_acc = [train_acc[i-1] for i in odd_epochs]
    odd_val_acc = [val_acc[i-1] for i in odd_epochs]

    # Vẽ biểu đồ
    plt.figure(figsize=(8, 6))
    plt.plot(odd_epochs, odd_train_acc, label='Độ chính xác của tập huấn luyện ', color='blue')
    plt.plot(odd_epochs, odd_val_acc, label='Độ chính xác của tập kiểm thử ', color='orange')
    plt.title('Độ Chính Xác Tập Huấn Luyện và Tập Kiểm Thử ')
    plt.xlabel('Epochs')
    plt.ylabel('Độ chính xác')
    plt.legend(loc='lower right')
    plt.grid(True)
    plt.show()

if __name__ == '__main__':
    history_save_path = '/content/models/history_efficientnet.pkl'

    # Tải lịch sử huấn luyện
    history = load_history(history_save_path)

    # Vẽ biểu đồ độ chính xác
    plot_accuracy(history)

```

```python
import os
import pickle
import matplotlib.pyplot as plt

def plot_accuracy_multiple_models(histories, model_names):
    """
    Vẽ biểu đồ độ chính xác của tập huấn luyện và kiểm thử theo các epoch lẻ cho nhiều mô hình.

    Parameters:
    - histories: List chứa lịch sử của các mô hình (dạng dict).
    - model_names: List tên các mô hình tương ứng.
    """
    if not histories or not model_names or len(histories) != len(model_names):
        print("Danh sách lịch sử hoặc tên mô hình không hợp lệ.")
        return

    plt.figure(figsize=(10, 8))

    for history, model_name in zip(histories, model_names):
        # Truy xuất độ chính xác của tập huấn luyện và kiểm thử
        train_acc = history.get('accuracy', [])
        val_acc = history.get('val_accuracy', [])

        # Kiểm tra dữ liệu
        if not train_acc or not val_acc:
            print(f"Không có dữ liệu độ chính xác cho mô hình {model_name}.")
            continue

        # Lọc các epoch lẻ
        epochs = range(1, len(train_acc) + 1)
        odd_epochs = [epoch for epoch in epochs if epoch % 2 != 0]

        # Lọc độ chính xác tương ứng với các epoch lẻ
        odd_train_acc = [train_acc[i - 1] for i in odd_epochs]
        odd_val_acc = [val_acc[i - 1] for i in odd_epochs]

        # Vẽ biểu đồ cho mô hình hiện tại
        plt.plot(odd_epochs, odd_train_acc, label=f'{model_name} - Train', linestyle='--')
        plt.plot(odd_epochs, odd_val_acc, label=f'{model_name} - Validation')

    # Định dạng biểu đồ
    plt.title('So sánh độ chính xác của các mô hình theo Epoch')
    plt.xlabel('Epochs')
    plt.ylabel('Độ chính xác')
    plt.legend(loc='lower right')
    plt.grid(True)
    plt.show()

if __name__ == '__main__':
    # Đường dẫn tới lịch sử của các mô hình
    model_paths = [
        '/content/models/history_mobilenetv2.pkl',
        '/content/models/history_clcm.pkl',
        '/content/models/history_resnet.pkl',
        '/content/models/history_vgg.pkl'
    ]

    # Tên các mô hình
    model_names = ['mobilenetv2', 'clcm', 'resnet','vgg']

    # Tải lịch sử của tất cả các mô hình
    histories = [load_history(path) for path in model_paths]

    # Vẽ biểu đồ so sánh
    plot_accuracy_multiple_models(histories, model_names)
```
